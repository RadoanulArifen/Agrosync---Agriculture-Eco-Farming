```python

#Clean & Structure the data
def clean_data(data):
    text_to_num = {"one":1, "two":2, "three":3, "four":4, "five":5}
    cleaned_data = []
    unique_users = set()

    for user in data:
        #clean ratings
        raw_ratting = str( user["rating"]).strip().lower()
        if(raw_ratting in text_to_num):
            raw_ratting = text_to_num[raw_ratting]

        user["rating"] = raw_ratting

        #Handling missing value
        raw_age = user.get("age")
        if(raw_age == None):
            user["age"] = None

        #Deduplication
        if(user["name"].strip() in unique_users):
            continue
        unique_users.add(user["name"])
        cleaned_data.append(user)

    return cleaned_data
```

```python
clean_data(data)
```

```python
#get meaningful insights

def get_insights(data):
  #avg rating
  tot_rating = 0
  for user in data:
    tot_rating += float(user["rating"])
  avg_rating = tot_rating / len(data)
  print(f"Average rating: {avg_rating}")

  #percentage of users with poor rating
  num_poor_rating = 0
  for user in data:
    if(float(user["rating"]) < 3):
      num_poor_rating += 1
  print(f"Percentage of users with poor rating: {num_poor_rating / len(data) * 100}%")
```

```python
# Re-running the insights function using the corrected code
cleaned_data = clean_data(data)
get_insights(cleaned_data)
```

```python
#Build the recommendation features

def get_recommendations(data):
  recommendation = []

  for user in data:
    curr_recomm = {}
    curr_recomm["name"] = user["name"]

    if(float(user["rating"]) >=4):
      curr_recomm["brand"] = "Apple"
    else:
      curr_recomm["brand"] = "Samsung"

    recommendation.append(curr_recomm)
  return recommendation
```
